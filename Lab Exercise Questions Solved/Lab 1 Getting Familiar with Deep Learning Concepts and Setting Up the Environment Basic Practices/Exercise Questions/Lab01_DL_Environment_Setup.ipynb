```python
# Import Libraries
import tensorflow as tf
import numpy as np

print("TensorFlow version:", tf.__version__)
print("NumPy version:", np.__version__)
print("GPU Available:", tf.config.list_physical_devices('GPU'))
```

```python
# Creating tensors of different ranks
scalar = tf.constant(7)
vector = tf.constant([1, 2, 3])
matrix = tf.constant([[1, 2], [3, 4]])

print("Scalar:", scalar)
print("Vector:", vector)
print("Matrix:\n", matrix)
print("Matrix shape:", matrix.shape)
```

```python
# Arithmetic and matrix operations
a = tf.constant([[1, 2], [3, 4]])
b = tf.constant([[5, 6], [7, 8]])

print("Addition:\n", a + b)
print("Element-wise multiplication:\n", a * b)
print("Matrix multiplication:\n", tf.matmul(a, b))
```

```python
# Reshaping and indexing
v = tf.constant([10, 20, 30, 40, 50, 60])
reshaped = tf.reshape(v, (2, 3))
print("Reshaped (2x3):\n", reshaped)
print("First row:", reshaped[0])
#print("Element at row 1, col 2:", reshaped[1, 2])
```

```python

# Converting between NumPy and TensorFlow
np_array = np.array([1.0, 2.0, 3.0])
tensor_from_np = tf.constant(np_array)
back_to_np = tensor_from_np.numpy()
print("NumPy -> Tensor:", tensor_from_np)
print("Tensor -> NumPy:", back_to_np)
```

```python
import matplotlib.pyplot as plt

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Training images shape:", x_train.shape)
print("Training labels shape:", y_train.shape)
print("Number of test images:", x_test.shape[0])

plt.imshow(x_train[0], cmap="gray")
plt.title(f"Label: {y_train[0]}")
plt.show()
```

```python
import numpy as np

# Inputs, weights and bias
x = np.array([0.5, 0.8, 0.2])
w = np.array([0.4, -0.6, 0.9])
b = 0.1

# Weighted sum
z = np.dot(x, w) + b

# Sigmoid activation
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

a = sigmoid(z)
print("Weighted sum (z):", z)
print("Neuron output (a):", a)
```

```python
import numpy as np

x = np.array([1.0, 0.3, -0.5])
w = np.array([0.2, 0.8, -0.1])
b = 0.05

z = np.dot(x, w) + b
relu_out = max(0.0, z)

print(f"x dot w: {np.dot(x, w)}")
print(f"z: {z}")
print(f"ReLU(z): {relu_out}")
```
